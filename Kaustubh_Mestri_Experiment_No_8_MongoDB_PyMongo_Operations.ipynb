```python
db = client["employee_db"]
collection = db["employees"]

print("Database:", db.name)
print("Collection:", collection.name)
```

```python
print("Number of documents:", collection.count_documents({}))
```

```python
collection.delete_many({})

print("Collection cleared")
```

```python
employees = [
    {
        "emp_id": 101,
        "name": "Amit",
        "department": "IT",
        "salary": 50000
    },
    {
        "emp_id": 102,
        "name": "Pooja",
        "department": "HR",
        "salary": 40000
    },
    {
        "emp_id": 103,
        "name": "Rahul",
        "department": "IT",
        "salary": 60000
    },
    {
        "emp_id": 104,
        "name": "Neha",
        "department": "Finance",
        "salary": 55000
    },
    {
        "emp_id": 105,
        "name": "Rohan",
        "department": "IT",
        "salary": 45000
    }
]

result = collection.insert_many(employees)

print("Inserted documents:", len(result.inserted_ids))
```

Search documents

```python
for doc in collection.find():
    print(doc)
```

Search for one employee

```python
result = collection.find_one({"emp_id": 103})

print(result)
```

Search using a condition

```python
results = collection.find({"salary": {"$gt": 50000}})

for doc in results:
    print(doc)
```

Update a document

```python
result = collection.update_one(
    {"emp_id": 101},
    {"$set": {"salary": 52000}}
)

print("Documents updated:", result.modified_count)
```

```python
print(collection.find_one({"emp_id": 101}))
```

Update multiple documents

```python
result = collection.update_many(
    {"department": "IT"},
    {"$inc": {"salary": 1000}}
)

print("Documents updated:", result.modified_count)
```

Replace a document

```python
result = collection.replace_one(
    {"emp_id": 102},
    {
        "emp_id": 102,
        "name": "Pooja",
        "department": "HR",
        "salary": 45000
    }
)

print("Documents replaced:", result.modified_count)
```

```python
print(collection.find_one({"emp_id": 102}))
```

Sort documents

```python
results = collection.find().sort("salary", -1)

for doc in results:
    print(doc["name"], doc["salary"])
```

For ascending order

```python
results = collection.find().sort("salary", 1)
```

count documents

```python
count = collection.count_documents({})

print("Total employees:", count)
```

Aggregation

```python
pipeline = [
    {
        "$group": {
            "_id": "$department",
            "average_salary": {"$avg": "$salary"}
        }
    }
]

results = collection.aggregate(pipeline)

for doc in results:
    print(doc)
```

aggregation: total salary

```python
pipeline = [
    {
        "$group": {
            "_id": "$department",
            "total_salary": {"$sum": "$salary"}
        }
    }
]

results = collection.aggregate(pipeline)

for doc in results:
    print(doc)
```

Removing a document

```python
 result = collection.delete_one({"emp_id": 105})

print("Documents deleted:", result.deleted_count)
```

Final verification

```python
 print("Final document count:", collection.count_documents({}))
```

```python
for doc in collection.find():
    print(doc)
```

```python
 print("===== FINAL COLLECTION =====")

for doc in collection.find():
    print(
        "ID:", doc["emp_id"],
        "| Name:", doc["name"],
        "| Department:", doc["department"],
        "| Salary:", doc["salary"]
    )

print("\nTotal Documents:", collection.count_documents({}))
```
