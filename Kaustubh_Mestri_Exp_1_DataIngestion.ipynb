**Aim**

To perform basic data ingestion by creating, reading, merging CSV files in Python.

**Theory**

Data ingestion is the process of collecting, importing, and transferring data from one or more sources into a system for storage, processing, and analysis. It is the first and one of the most important stages of the data engineering pipeline, as it ensures that data is available in a structured format for further operations. CSV (Comma-Separated Values) is a widely used file format for storing tabular data because it is simple, lightweight, and compatible with many applications.

In this experiment, the Python Pandas library is used to perform basic data ingestion tasks. Initially, two separate CSV files are created and stored. These files are then read into Pandas DataFrames using the read_csv() function. After reading the data, both DataFrames are combined into a single DataFrame using the concat() function. Finally, the merged data is saved as a new CSV file using the to_csv() function. This experiment demonstrates the basic workflow of reading, writing, merging, and storing data, which forms the foundation of a simple data pipeline in data engineering.

```python
import pandas as pd

data2 = {
    "ID":[4,5,6],
    "Name":["Sneha","Rohan","Kiran"],
    "Marks":[85,88,91]
}

df2 = pd.DataFrame(data2)

df2.to_csv("student2.csv",index=False)
```

```python
import pandas as pd

df1 = pd.read_csv("student1.csv")

df2 = pd.read_csv("student2.csv")

print("Student 1 Data:")
print(df1)

print("\nStudent 2 Data:")
print(df2)
```

```python
merged_df = pd.concat([df1, df2], ignore_index=True)

print("Merged Data:")
print(merged_df)
```

```python
merged_df.to_csv("Merged_Students.csv", index=False)

print("Merged CSV file saved successfully!")
```

**Conclusion**

The experiment was successfully completed by creating two CSV files, reading them into Pandas DataFrames, merging the data using the concat() function, and storing the combined data as a new CSV file. This experiment demonstrated the basic concepts of data ingestion and showed how Pandas can efficiently handle reading, writing, and processing tabular data, which is an essential step in the data engineering pipeline.

**Activity 2: Analysis of an Open-Source Dataset**


** Aim**

To load and analyze an open-source dataset using the Pandas library in Python by examining its structure, dimensions, column names, data types, missing values, and statistical information.

**Theory**

Dataset analysis is an important step in data visualization and machine learning. It helps in understanding the structure and quality of the data before performing further analysis or building predictive models.

In this experiment, the Breast Cancer Wisconsin Dataset, an open-source dataset available in the scikit-learn library, was used. The dataset contains various medical measurements of breast cancer cell nuclei, such as radius, texture, perimeter, area, and smoothness. Each record is classified as either malignant (cancerous) or benign (non-cancerous) using the target column.

Using the Pandas library, the dataset is loaded into a DataFrame and analyzed by determining the number of rows and columns, displaying column names, checking data types, identifying missing values, and generating statistical summaries. This analysis helps in understanding the dataset before applying visualization or machine learning techniques.


```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

# Load the open-source dataset
dataset = load_breast_cancer()

# Convert it into a Pandas DataFrame
df = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names
)

# Add the target column
df["target"] = dataset.target

print("Dataset loaded successfully!")
display(df.head())
```

```python
print("Column Names:")

for column in df.columns:
    print(column)
```

```python
print("Data Types of All Columns:")

print(df.dtypes)
```

```python
print("Complete Dataset Information:")

df.info()
```

```python
print("Statistical Summary:")

display(df.describe())
```

```python
duplicate_rows = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_rows)
```

```python
df.to_csv("Breast_Cancer_Dataset.csv", index=False)

print("Open-source dataset saved successfully!")
```

**Conclusion**

The experiment was successfully completed by loading and analyzing the Breast Cancer Wisconsin Dataset using Python and the Pandas library. The dataset's structure, dimensions, column names, data types, missing values, and statistical summary were examined successfully. This activity improved the understanding of data exploration and demonstrated the importance of analyzing a dataset before performing data visualization or machine learning tasks.

**Activity 3: Automatic File Selection and Loading**

**Aim **

To develop a Python program that allows the user to automatically select and load CSV, Excel, or JSON files for data analysis without manually specifying the file path.

**Theory**

In data analysis, datasets are often stored in different file formats such as CSV, Excel, and JSON. Instead of manually entering the file path in the program, Python provides file selection utilities that allow users to browse and select a file interactively.

In this experiment, Google Colab's files.upload() function is used to upload a dataset. The program automatically identifies the uploaded file's extension and loads it using the appropriate Pandas function, such as read_csv(), read_excel(), or read_json(). This approach makes the program more flexible, user-friendly, and capable of handling multiple file formats with a single implementation.

```python
from google.colab import files
import pandas as pd
import io
import os

# Open file selection window
uploaded = files.upload()

# Get the selected filename
file_name = list(uploaded.keys())[0]

print("\nSelected File:", file_name)

# Find the extension of the selected file
file_extension = os.path.splitext(file_name)[1].lower()

# Automatically load the file
if file_extension == ".csv":
    selected_df = pd.read_csv(
        io.BytesIO(uploaded[file_name])
    )

elif file_extension in [".xlsx", ".xls"]:
    selected_df = pd.read_excel(
        io.BytesIO(uploaded[file_name])
    )

elif file_extension == ".json":
    try:
        selected_df = pd.read_json(
            io.BytesIO(uploaded[file_name])
        )
    except ValueError:
        selected_df = pd.read_json(
            io.BytesIO(uploaded[file_name]),
            lines=True
        )

else:
    raise ValueError(
        "Unsupported file format. Please upload CSV, Excel or JSON."
    )

print("\nFile loaded successfully!")

print("\nFirst Five Rows:")
display(selected_df.head())
```

```python
print("Number of Rows:", selected_df.shape[0])
print("Number of Columns:", selected_df.shape[1])

print("\nColumn Names:")
print(selected_df.columns.tolist())

print("\nData Types:")
print(selected_df.dtypes)

print("\nMissing Values:")
print(selected_df.isnull().sum())

print("\nTotal Missing Values:")
print(selected_df.isnull().sum().sum())

print("\nDataset Information:")
selected_df.info()
```
