### Theory

MongoDB is a NoSQL, document-oriented database that stores data in flexible JSON-like documents instead of traditional rows and columns. In a Hospital Management System, MongoDB can be used to store information about departments, doctors, patients, appointments, and medical reports in separate collections. Each collection contains documents representing individual records.

PyMongo is a Python library that provides an interface for connecting Python applications with MongoDB. It allows developers to create databases and collections and perform various database operations programmatically. In this experiment, Google Colab is used as the Python environment and MongoDB Atlas is used as the cloud-based MongoDB database.

The main database operations performed are CRUD operations: Create, Read, Update, and Delete. Documents can be inserted using `insert_one()` or `insert_many()`, retrieved using `find()` or `find_one()`, modified using `update_one()` or `update_many()`, and removed using `delete_one()` or `delete_many()`. The `replace_one()` operation can be used to replace an entire document.

MongoDB also provides operations such as sorting and counting for organizing and analyzing records. The `aggregate()` function is used to process multiple documents and generate summarized information such as counts, averages, and grouped results. These features make MongoDB suitable for managing and retrieving hospital-related information efficiently.

In this experiment, a Python-based Hospital Management System is implemented using MongoDB and PyMongo, demonstrating how NoSQL database operations can be integrated into a real-world healthcare application.

Create Database

```python
departments = db["departments"]
doctors = db["doctors"]
patients = db["patients"]
appointments = db["appointments"]
medical_reports = db["medical_reports"]
```

Clear Previous Data

```python
departments.delete_many({})
doctors.delete_many({})
patients.delete_many({})
appointments.delete_many({})
medical_reports.delete_many({})
```

Insert Departments

```python
department_id = input("Enter Department ID: ")
name = input("Enter Department Name: ")
location = input("Enter Department Location: ")

department_data = {
    "department_id": department_id,
    "name": name,
    "location": location
}

departments.insert_one(department_data)

print("Department inserted successfully!")
```

Insert Doctors

```python
doctor_id = input("Enter Doctor ID: ")
name = input("Enter Doctor Name: ")
specialization = input("Enter Specialization: ")
department_id = input("Enter Department ID: ")
phone = input("Enter Phone Number: ")

doctor_data = {
    "doctor_id": doctor_id,
    "name": name,
    "specialization": specialization,
    "department_id": department_id,
    "phone": phone
}

doctors.insert_one(doctor_data)

print("Doctor inserted successfully!")
```

Insert Patients

```python
patient_id = input("Enter Patient ID: ")

while not patient_id.isalnum():
    print("Invalid Patient ID! ID must contain only letters and numbers.")
    patient_id = input("Enter Patient ID again: ")

name = input("Enter Patient Name: ")
age = int(input("Enter Patient Age: "))
gender = input("Enter Gender: ")
phone = input("Enter Phone Number: ")

patient_data = {
    "patient_id": patient_id,
    "name": name,
    "age": age,
    "gender": gender,
    "phone": phone
}

patients.insert_one(patient_data)

print("Patient inserted successfully!")
```

Insert Appointments

```python
appointment_id = input("Enter Appointment ID: ")
patient_id = input("Enter Patient ID: ")
doctor_id = input("Enter Doctor ID: ")
date = input("Enter Appointment Date (YYYY-MM-DD): ")
time = input("Enter Appointment Time: ")
status = input("Enter Appointment Status: ")

appointment_data = {
    "appointment_id": appointment_id,
    "patient_id": patient_id,
    "doctor_id": doctor_id,
    "date": date,
    "time": time,
    "status": status
}

appointments.insert_one(appointment_data)

print("Appointment inserted successfully!")
```

Insert Medical Reports

```python
report_id = input("Enter Report ID: ")
patient_id = input("Enter Patient ID: ")
doctor_id = input("Enter Doctor ID: ")
diagnosis = input("Enter Diagnosis: ")
prescription = input("Enter Prescription: ")
report_date = input("Enter Report Date (YYYY-MM-DD): ")

report_data = {
    "report_id": report_id,
    "patient_id": patient_id,
    "doctor_id": doctor_id,
    "diagnosis": diagnosis,
    "prescription": prescription,
    "report_date": report_date
}

medical_reports.insert_one(report_data)

print("Medical report inserted successfully!")
```

Display all patients

```python
for patient in patients.find():
    pprint(patient)
```

Find one patient

```python
pprint(patients.find_one({"patient_id": 201}))
```


Find patients older than 30

```python
for patient in patients.find({"age": {"$gt": 30}}):
    pprint(patient)
```

Update a Patient

```python
patient_id = input("Enter Patient ID to update: ")
new_age = int(input("Enter New Age: "))

patients.update_one(
    {"patient_id": patient_id},
    {"$set": {"age": new_age}}
)

print("Patient updated successfully!")
```

```python
pprint(patients.find_one({"patient_id": 201}))
```

Update Appointment

```python
appointments.update_one(
    {"appointment_id": 301},
    {"$set": {"status": "Completed"}}
)

print("Appointment updated successfully")
```

Replace a Medical Report

```python
medical_reports.replace_one(
    {"report_id": 401},
    {
        "report_id": 401,
        "patient_id": 201,
        "doctor_id": 101,
        "diagnosis": "Controlled Blood Pressure",
        "treatment": "Continue Medication"
    }
)

print("Medical report replaced successfully")
```

Sort by age ascending

```python
for patient in patients.find().sort("age", 1):
    pprint(patient)
```

Sort by age descending

```python
for patient in patients.find().sort("age", -1):
    pprint(patient)
```

Count total patients:

```python
print("Total patients:", patients.count_documents({}))
```

Count total doctors:

```python
print("Total doctors:", doctors.count_documents({}))
```

Count scheduled appointments:

```python
print(
    "Scheduled appointments:",
    appointments.count_documents({"status": "Scheduled"})
)
```

Delete a Patient

```python
patient_id = input("Enter Patient ID to delete: ")

patients.delete_one(
    {"patient_id": patient_id}
)

print("Patient deleted successfully!")
```

```python
for patient in patients.find():
    pprint(patient)
```

Delete an Appointment

```python
appointments.delete_one({"appointment_id": 303})

print("Appointment deleted successfully")
```

Aggregation

```python
result = patients.aggregate([
    {
        "$group": {
            "_id": None,
            "average_age": {"$avg": "$age"}
        }
    }
])

for r in result:
    pprint(r)
```

Another Aggregation — Patients by Gender

```python
result = patients.aggregate([
    {
        "$group": {
            "_id": "$gender",
            "total": {"$sum": 1}
        }
    }
])

for r in result:
    pprint(r)
```
