```python
import psycopg2
from sqlalchemy import create_engine
```

```python
conn = psycopg2.connect(
    dbname="breast_cancer_db",
    user="postgres",
    password="postgres",
    host="localhost",
    port="5432"
)

print("Connected to PostgreSQL successfully!")
```

```python
engine = create_engine(
    "postgresql+psycopg2://postgres:postgres@localhost:5432/breast_cancer_db"
)

print("SQLAlchemy engine created successfully!")
```

```python
df.to_sql(
    "breast_cancer",
    engine,
    if_exists="replace",
    index=False
)

print("Dataset successfully loaded into PostgreSQL!")
print("Total records loaded:", len(df))
```

```python
query = """
SELECT *
FROM breast_cancer
LIMIT 5;
"""

result = pd.read_sql_query(query, engine)

display(result)
```

```python
query = """
SELECT COUNT(*) AS total_records
FROM breast_cancer;
"""

result = pd.read_sql_query(query, engine)

display(result)
```
