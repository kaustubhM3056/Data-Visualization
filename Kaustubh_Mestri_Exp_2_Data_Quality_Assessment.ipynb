**Aim**

To perform comprehensive data quality assessment on a structured dataset by identifying and correcting data quality issues such as missing values, duplicate records, inconsistent data formats, invalid entries, and outliers using Python and the Pandas library.

**Theory**

Data quality assessment is the process of evaluating the accuracy, completeness, consistency, validity, and reliability of a dataset before performing data analysis or visualization. Poor-quality data may contain missing values, duplicate records, inconsistent formats, invalid entries, or outliers, which can lead to incorrect results and poor decision-making.

Data cleaning is the process of detecting and correcting these issues to improve the quality of the dataset. The Pandas library in Python provides several functions such as isnull(), duplicated(), fillna(), drop_duplicates(), and data type conversion methods to efficiently clean and preprocess data.

After cleaning, Matplotlib is used to visualize the processed data through different charts such as bar charts, line charts, pie charts, histograms, and scatter plots. Data visualization helps in understanding patterns, trends, distributions, and relationships within the dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "ID": [1, 2, 3, 4, 5, 5],
    "Name": ["Rahul", "Priya", "Amit", "Sneha", None, "Rohan"],
    "Age": [20, 21, None, 22, 20, 20],
    "Marks": [80, 90, 75, None, 88, 88]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python
print("Dataset Information:")
df.info()
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python
display(df.describe())
```

Replace the missing values.

```python
df["Name"] = df["Name"].fillna("Unknown")

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

Display the cleaned dataset:

```python
print("Dataset after filling missing values:")
display(df)
```

Remove duplicate rows

```python
df = df.drop_duplicates()

print("Dataset after removing duplicates:")
display(df)
```

Verify the cleaned dataset

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

Bar Chart

```python
plt.figure(figsize=(6,4))

plt.bar(df["Name"], df["Marks"])

plt.title("Student Marks")

plt.xlabel("Student Name")

plt.ylabel("Marks")

plt.show()
```

Line Chart


```python
plt.figure(figsize=(6,4))

plt.plot(df["Name"], df["Marks"], marker="o")

plt.title("Student Marks")

plt.xlabel("Student Name")

plt.ylabel("Marks")

plt.show()
```

Pie Chart

```python
plt.figure(figsize=(6,6))

plt.pie(df["Marks"],
        labels=df["Name"],
        autopct="%1.1f%%")

plt.title("Marks Distribution")

plt.show()
```

Histogram

```python
plt.figure(figsize=(6,4))

plt.hist(df["Marks"], bins=5)

plt.title("Histogram of Marks")

plt.xlabel("Marks")

plt.ylabel("Frequency")

plt.show()
```

Scatter Plot

```python
plt.figure(figsize=(6,4))

plt.scatter(df["Age"], df["Marks"])

plt.title("Age vs Marks")

plt.xlabel("Age")

plt.ylabel("Marks")

plt.show()
```

```python
df.to_csv("Cleaned_Dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
